# Python 聚餐費用試算器完整參考解答

教師參考版。使用變數、字串處理、數字、布林與基本格式化，依題目規定先計算服務費，再扣折價券。

以下三組測試各自包含完整的輸入、整理、計算與輸出，可各自獨立執行。輸出已由 Python 實際執行產生。

## 測試一：原始資料

本組人數為 4 人，折價券為 100 元。請執行下方完整儲存格。

````python
# 1. 原始資料：保留文字資料，另外建立整理後的變數。
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 整理聚餐名稱。
clean_event_name = event_name.strip().lower()

# 3. 清理餐費文字，再轉換餐費與人數的型別。
clean_meal_text = meal_amount.replace("NT$", "").replace(",", "").replace(" ", "")
meal_value = float(clean_meal_text)
people_count = int(people)

# 4. 先按原始餐費計算服務費，再扣除折價券。
service_fee = meal_value * service_rate
total_amount = meal_value + service_fee - discount
amount_per_person = total_amount / people_count

# 5. 比較結果直接保存為 bool。等於預算時不算超過。
is_over_budget = amount_per_person > budget_per_person

# 6. 輸出時才將金額格式化為兩位小數。
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_value:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{amount_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
````

## 程式逐步說明

### 1. 名稱整理

`event_name.strip().lower()` 先去掉頭尾空白，再轉成小寫，得到 `"weekend dinner"`。字串方法回傳新字串，因此用 `clean_event_name` 接住結果，原始 `event_name` 保持不變。

### 2. 餐費清理與型別轉換

三次 `replace()` 分別移除 `NT$`、逗號及半形空白，得到文字 `"1200"`。`float()` 將它轉成數值 `1200.0`，可以參與乘除計算。人數則用 `int(people)` 轉成整數。

本題使用題目提供的有效資料，不另外處理其他貨幣符號、無效輸入或零人分攤。

### 3. 計算順序

服務費以原始餐費計算，所以三組測試的服務費都是 `1200 × 0.10 = 120` 元。算出餐費加服務費後，再扣除該組折價券，最後除以人數。

### 4. 布林比較

`amount_per_person > budget_per_person` 直接產生 `True` 或 `False`，因此這題可以不使用 `if`。題目問的是「超過」，所以使用 `>`，每人費用等於預算時結果是 `False`。

### 5. 輸出格式

f-string 中的 `:.2f` 讓金額以小數點後兩位顯示。例如 `305.0` 顯示成 `305.00`。格式化產生顯示文字，原本的數值變數仍可繼續計算。本題保留原數值完成計算，再於最後輸出時格式化。


## 兩題問答參考答案

### 問題一：為什麼 `"4"` 要先轉成數字？

`"4"` 是 `str`，`4` 才是 `int`。本題需要使用人數作為除數，計算每人分攤費用。Python 無法直接將餐費數值除以字串，因此先用 `int(people)` 得到整數 `4`，再參與計算。

文字外觀看起來像數字，並不代表 Python 已經將它當成數值。例如 `"4" + "1"` 會得到 `"41"`，而 `4 + 1` 才會得到 `5`。

### 問題二：如果先扣折價券，再計算服務費？

第一組先扣折價券，餐費基準會變成：

```text
1,200 − 100 = 1,100 元
```

再以 1,100 元計算 10% 服務費：

```text
服務費 = 1,100 × 0.10 = 110 元
應付總額 = 1,100 + 110 = 1,210 元
每人費用 = 1,210 ÷ 4 = 302.50 元
```

題目規定的計算則是：

```text
服務費 = 1,200 × 0.10 = 120 元
應付總額 = 1,200 + 120 − 100 = 1,220 元
每人費用 = 1,220 ÷ 4 = 305.00 元
```

兩種順序的服務費基準不同，相差 `100 × 0.10 = 10` 元。先折價再收服務費會少收 10 元，因此本題必須依題目規定，以原始餐費計算服務費。
